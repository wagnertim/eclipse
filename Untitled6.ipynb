```python
import numpy as np
import netCDF4
import matplotlib.pyplot as plt
```

```python
e12f = '/Users/timw/data/ecor/sgp30ecorE14.b1.20120521.000000.cdf'
e17f = '/Users/timw/data/ecor/sgp30ecorE14.b1.20170821.000000.cdf'
e23f = '/Users/timw/data/ecor/sgpecorsfE14.b1.20231014.000000.nc'
e24f = '/Users/timw/data/ecor/sgpecorsfE14.b1.20240408.000000.nc'

ds12 = netCDF4.Dataset(e12f)
ds17 = netCDF4.Dataset(e17f)
ds23 = netCDF4.Dataset(e23f)
ds24 = netCDF4.Dataset(e24f)


class e12:

    time = ds12['time'][:]/3600
    h = ds12['h'][:]
    lv_e = ds12['lv_e'][:]



class e17:

    time = ds17['time'][:]/3600
    h = ds17['h'][:]
    lv_e = ds17['lv_e'][:]
    temp = ds17['mean_t'][:]
    q = ds17['mean_q'][:]/1000
    wspd = ds17['wind_spd'][:]


class e24:

    time = ds24['time'][:]/3600
    h = ds24['sensible_heat_flux'][:]
    lv_e = ds24['latent_flux'][:]
    temp = ds24['air_temperature'][:]
    q = ds24['h2o_mixing_ratio'][:]
    wspd = ds24['mean_wind'][:]



s17f = '/Users/timw/data/sgpmet/sgpmetE13.b1.20170821.000000.cdf'
s24f = '/Users/timw/data/sgpmet/sgpmetE13.b1.20240408.000000.cdf'

ds17s = netCDF4.Dataset(s17f)
ds24s = netCDF4.Dataset(s24f)

class s17:
    time = ds17s['time'][:]/3600
    temp = ds17s['temp_mean'][:]
    wspd = ds17s['wspd_arith_mean'][:]
    pres = ds17s['atmos_pressure'][:]
    w = ds17s['vapor_pressure_mean'][:]

class s24:
    time = ds24s['time'][:]/3600
    temp = ds24s['temp_mean'][:]
    wspd = ds24s['wspd_arith_mean'][:]
    pres = ds24s['atmos_pressure'][:] 
    w = ds24s['vapor_pressure_mean'][:]

s17.q = (0.622 * s17.w) / (s17.pres - s17.w) 
s24.q = (0.622 * s24.w) / (s24.pres - s24.w) 
```

```python


fig, ((ax1, ax2, ax3), (ax4, ax5, ax6)) = plt.subplots(nrows=2, ncols=3, figsize=(14.8, 9))




ax1.plot(e17.time-(18+4/60), e17.h,label='2017',linewidth=2)
ax1.plot(e24.time-(18+47/60), e24.h,label='2014',linewidth=2)
ax1.set_xlim(-3,3)
ax1.set_ylim(-80,280)
ax1.hlines(y=0, xmin=-3, xmax=3,color='k',linewidth=3)

ax1.grid()
ax1.set_ylabel('Flux [W m$^2$]')
ax1.set_title('a. Sensible Heat Flux')
ax1.legend(loc='lower right')


ax2.plot(e17.time-(18+4/60), e17.lv_e,label='2017',linewidth=2)
ax2.plot(e24.time-(18+47/60), e24.lv_e,label='2024',linewidth=2)
ax2.set_xlim((-3,3))
ax2.set_ylim(-80,280)
ax2.hlines(y=0, xmin=-3, xmax=3,color='k',linewidth=3)

ax2.grid()
ax2.set_ylabel('Flux [W m$^2$]')
ax2.set_title('b. Latent Heat Flux')
ax2.legend(loc='lower right')



ax3.plot(e17.time-(18+4/60), e17.h/e17.lv_e,label='2017',linewidth=2)
ax3.plot(e24.time-(18+47/60),e24.h/e24.lv_e,label='2024',linewidth=2)
ax3.set_xlim((-3,3))
ax3.set_ylim(-1,2)
ax3.hlines(y=0, xmin=-3, xmax=3,color='k',linewidth=3)

ax3.grid()
ax3.set_ylabel('Bowen Ratio')
ax3.set_title('c. Bowen Ratio')
ax3.legend(loc='lower right')



ax4.plot(s17.time-(18+4/60), s17.temp,label='2017',linewidth=2)
ax4.plot(s24.time-(18+47/60), s24.temp,label='2024',linewidth=2)
ax4.set_xlim((-3,3))
ax4.grid()
ax4.set_title('d. Temperature')
ax4.set_xlabel('Time [hours relative to max eclipse]')
ax4.set_ylabel('Temperature [C]')
ax4.legend(loc='lower right')

ax5.plot(s17.time-(18+4/60), s17.q*1000,label='2017',linewidth=2)
ax5.plot(s24.time-(18+47/60), s24.q*1000,label='2024',linewidth=2)
ax5.set_xlim((-3,3))
ax5.grid()
ax5.set_title('e. Mixing Ratio')
ax5.set_xlabel('Time [hours relative to max eclipse]')
ax5.set_ylabel('Mixing Ratio [g kg$^{-1}$]')
ax5.legend(loc='lower right')
ax5.set_ylim(0,17)

ax6.plot(s17.time-(18+4/60), s17.wspd,label='2017',linewidth=2)
ax6.plot(s24.time-(18+47/60), s24.wspd,label='2024',linewidth=2)
ax6.set_xlim((-3,3))
ax6.grid()
ax6.set_title('f. Wind Speed')
ax6.set_xlabel('Time [hours relative to max eclipse]')
ax6.set_ylabel('Wind Speed [m s$^{-1}$]')
ax6.legend(loc='lower right')

plt.savefig('fluxes_and_bowen_ratio.png')
```
