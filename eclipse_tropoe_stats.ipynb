Define a couple of functions, one to read the TROPoe output and the other to find the nearest index.

```python
def readTROPoe(f):

    ds = netCDF4.Dataset(f)
    
    class out:
        
        unixTime = ds['base_time'][:] + ds['time_offset'][:]
        time = ds['hour'][:] # Hours UTC 
        height = ds['range'][:]
        temp = ds['temperature'][:]
        wv = ds['waterVapor'][:]
        theta = ds['theta'][:]
        back = ds['attenuated_backscatter'][:]
    
    return out

def findNearest(array,value):
    idx = (np.abs(array-value)).argmin()
    return idx  
```
